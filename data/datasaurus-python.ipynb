## 1. 読み込む / Load

`dataset`はデータの名前、`x`と`y`は1つの点の座標です。
`dataset` names the group; `x` and `y` give the coordinates of one point.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Colabの左のファイル一覧にCSVをアップロード / Upload the CSV in Colab's Files sidebar
df = pd.read_csv('datasaurus.csv')
assert {'dataset', 'x', 'y'}.issubset(df.columns)
df.head()
```

```python
# データ名と点の数 / Dataset names and point counts
df.groupby('dataset').size().rename('points')
```

## 2. 描く / Draw

まず`dino`を描きます。次に`selected_dataset`・`point_color`・`point_size`を変えて実行してみましょう。
Draw `dino` first. Then change the dataset, color, or size and run again.

```python
selected_dataset = 'dino'
point_color = '#b33b50'
point_size = 25

points = df[df['dataset'] == selected_dataset]
assert not points.empty, 'Choose a dataset name from the list above.'
fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(points['x'], points['y'], s=point_size, color=point_color, alpha=0.8)
ax.set(xlabel='x', ylabel='y', title=selected_dataset, xlim=(0, 100), ylim=(0, 110))
ax.set_aspect('equal', adjustable='box')
plt.show()
```

## 3. 数字で比べる / Compare the statistics

平均・標準偏差・相関が似ていたら、形も似ているでしょうか？まず予想してください。
If the means, standard deviations, and correlations are similar, will the pictures be similar? Make a prediction.

```python
summary = df.groupby('dataset').agg(
    points=('x', 'size'),
    mean_x=('x', 'mean'), mean_y=('y', 'mean'),
    sd_x=('x', 'std'), sd_y=('y', 'std')
)
correlations = {name: group['x'].corr(group['y']) for name, group in df.groupby('dataset')}
summary['correlation'] = pd.Series(correlations)
summary.round(2)
```

## 4. 全て見る / See them all

比較のため、軸の範囲と縦横の縮尺をそろえます。
Use identical limits and equal aspect ratios for a fair comparison.

```python
names = sorted(df['dataset'].unique())
fig, axes = plt.subplots(4, 4, figsize=(13, 13))
for ax, name in zip(axes.flat, names):
    points = df[df['dataset'] == name]
    ax.scatter(points['x'], points['y'], s=8, color='#b33b50')
    ax.set(title=name, xlim=(0, 100), ylim=(0, 110))
    ax.set_aspect('equal', adjustable='box')
for ax in list(axes.flat)[len(names):]:
    ax.set_visible(False)
fig.tight_layout()
plt.show()
```
