```python
import pandas as pd
```

```python
countries_total = pd.read_csv('countries_total.csv')
income_per_person = pd.read_csv('income_per_person.csv')
population_total = pd.read_csv('population_total.csv')
life_expectancy_years = pd.read_csv('life_expectancy_years.csv')
```

```python
# First, let's examine the structure of each dataset
print("Countries Total:")
print(countries_total.head())
print("\nColumns:", countries_total.columns.tolist())
print("\nShape:", countries_total.shape)
```

```python
print("Income per Person:")
print(income_per_person.head())
print("\nColumns:", income_per_person.columns.tolist())
print("\nShape:", income_per_person.shape)
```

```python
print("Population Total:")
print(population_total.head())
print("\nColumns:", population_total.columns.tolist())
print("\nShape:", population_total.shape)
```

```python
print("Life Expectancy Years:")
print(life_expectancy_years.head())
print("\nColumns:", life_expectancy_years.columns.tolist())
print("\nShape:", life_expectancy_years.shape)
```

```python
# Function to melt dataframes that have years as columns
def melt_year_data(df, id_col='geo', value_name='value'):
    """
    Melt a dataframe where years are columns into long format
    """
    # Get year columns (assuming they are numeric strings)
    year_cols = [col for col in df.columns if col != id_col and col.isdigit()]
    
    # Melt the dataframe
    melted = df.melt(id_vars=[id_col], 
                     value_vars=year_cols,
                     var_name='Year',
                     value_name=value_name)
    
    # Convert Year to integer
    melted['Year'] = melted['Year'].astype(int)
    
    return melted
```

```python
# Transform each dataset to long format
print("Melting datasets...")

# Melt population data
population_long = melt_year_data(population_total, value_name='Population')
print(f"Population melted: {population_long.shape}")

# Melt income data  
income_long = melt_year_data(income_per_person, value_name='Income_per_person')
print(f"Income melted: {income_long.shape}")

# Melt life expectancy data
life_expectancy_long = melt_year_data(life_expectancy_years, value_name='Life_Expectancy')
print(f"Life expectancy melted: {life_expectancy_long.shape}")

# Countries data should already have the right structure (assuming it has geo and continent columns)
print(f"Countries data: {countries_total.shape}")
print("Countries columns:", countries_total.columns.tolist())
```

```python
# Let's examine the 'geo' values in the melted datasets
print("Sample of population_long 'geo' values:")
print(population_long['geo'].head(10))
print(f"Unique geo values in population_long: {population_long['geo'].nunique()}")
print("Sample unique geo values:", population_long['geo'].unique()[:10])

print("\nSample of countries_total relevant columns:")
print(countries_total[['name', 'alpha-2', 'alpha-3']].head())
```

```python
# Combine all datasets
print("Combining datasets...")

# Start with population data as the base
combined_df = population_long.copy()

# Merge with income data
combined_df = combined_df.merge(income_long, on=['geo', 'Year'], how='outer')

# Merge with life expectancy data
combined_df = combined_df.merge(life_expectancy_long, on=['geo', 'Year'], how='outer')

# Merge with countries data to get country names and continents
# The 'geo' column in our datasets contains country names, which matches the 'name' column in countries_total
combined_df = combined_df.merge(countries_total, left_on='geo', right_on='name', how='left')

print(f"Combined dataset shape: {combined_df.shape}")
print("Combined dataset columns:", combined_df.columns.tolist())
print("\nFirst few rows:")
print(combined_df.head())
```

```python
# Clean up and rename columns to match the requested format
# Target columns: Country, Year, Population, Continent, Region, Sub-region, Life Expectancy, Income per person

# First, let's see what columns we actually have and rename appropriately
print("Available columns:", combined_df.columns.tolist())

# Create a mapping dictionary for renaming columns
column_mapping = {}

# Look for country name column (could be 'name', 'country', etc.)
for col in combined_df.columns:
    if col.lower() in ['name', 'country', 'country_name']:
        column_mapping[col] = 'Country'
        break

# Look for continent column (but keep region separate)
for col in combined_df.columns:
    if col.lower() in ['continent']:
        column_mapping[col] = 'Continent'
        break

# Add mappings for region and sub-region
if 'region' in combined_df.columns:
    column_mapping['region'] = 'Region'
if 'sub-region' in combined_df.columns:
    column_mapping['sub-region'] = 'Sub-region'

# Add other mappings
if 'Population' not in combined_df.columns:
    column_mapping['Population'] = 'Population'
if 'Life_Expectancy' in combined_df.columns:
    column_mapping['Life_Expectancy'] = 'Life Expectancy'
if 'Income_per_person' in combined_df.columns:
    column_mapping['Income_per_person'] = 'Income per person'

print("Column mapping:", column_mapping)

# Rename columns
combined_df = combined_df.rename(columns=column_mapping)

# Select and reorder the final columns
final_columns = ['Country', 'Year', 'Population', 'Continent', 'Region', 'Sub-region', 'Life Expectancy', 'Income per person']
available_columns = [col for col in final_columns if col in combined_df.columns]

print("Available final columns:", available_columns)

# Create the final dataframe
final_df = combined_df[available_columns + ['geo']].copy()  # Keep geo for reference

print(f"\nFinal dataset shape: {final_df.shape}")
print("Final dataset columns:", final_df.columns.tolist())
print("\nFirst 10 rows of final dataset:")
print(final_df.head(10))
```

```python
# Data quality check
print("Data Quality Summary:")
print("=" * 50)

print(f"Total rows: {len(final_df):,}")
print(f"Unique countries: {final_df['geo'].nunique()}")
print(f"Year range: {final_df['Year'].min()} - {final_df['Year'].max()}")

print("\nMissing values per column:")
for col in final_df.columns:
    missing_count = final_df[col].isnull().sum()
    missing_pct = (missing_count / len(final_df)) * 100
    print(f"  {col}: {missing_count:,} ({missing_pct:.1f}%)")

print(f"\nSample of data for year 2020:")
sample_2020 = final_df[final_df['Year'] == 2020].head()
print(sample_2020)
```

```python
# Re-create the final dataframe with region and sub-region included
print("Re-creating final dataframe with region and sub-region...")

# Update the column mapping to include region and sub-region
updated_column_mapping = {
    'name': 'Country',
    'region': 'Region', 
    'sub-region': 'Sub-region',
    'Life_Expectancy': 'Life Expectancy',
    'Income_per_person': 'Income per person'
}

# Apply the updated mapping
combined_df_updated = combined_df.rename(columns=updated_column_mapping)

# Define the final columns including region and sub-region
final_columns_updated = ['Country', 'Year', 'Population', 'Region', 'Sub-region', 'Life Expectancy', 'Income per person']
available_columns_updated = [col for col in final_columns_updated if col in combined_df_updated.columns]

print("Updated available columns:", available_columns_updated)

# Create the updated final dataframe
final_df_updated = combined_df_updated[available_columns_updated + ['geo']].copy()

print(f"\nUpdated final dataset shape: {final_df_updated.shape}")
print("Updated final dataset columns:", final_df_updated.columns.tolist())
print("\nFirst 10 rows of updated final dataset:")
print(final_df_updated.head(10))
```

```python
final_df.to_csv('gapminder.csv', index=False)
```
