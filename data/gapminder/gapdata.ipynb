```python
import pandas as pd
```

```python
pop = pd.read_csv('pop.csv')
gdp = pd.read_csv('gdp.csv')
lex = pd.read_csv('lex.csv')
```

```python
pop.head()
```

```python
# Pivot pop to long format
pop_long = pop.melt(id_vars='country', var_name='year', value_name='population')

# Pivot gdp to long format
gdp_long = gdp.melt(id_vars='country', var_name='year', value_name='gdp')

# Pivot lex to long format
lex_long = lex.melt(id_vars='country', var_name='year', value_name='lex')
```

```python
pop_long.head()
```

```python
gap_final = pop_long.merge(gdp_long, on=['country', 'year'], how='outer') \
                    .merge(lex_long, on=['country', 'year'], how='outer')
```

```python
gap_final.head()
```

```python
def convert_text_number(val):
    if isinstance(val, str):
        val = val.strip()
        if val.endswith('k'):
            return int(float(val[:-1]) * 1_000)
        elif val.endswith('M'):
            return int(float(val[:-1]) * 1_000_000)
        else:
            try:
                return int(val.replace(',', ''))
            except ValueError:
                return pd.NA
    return val

pop_long['population'] = pop_long['population'].apply(convert_text_number).astype('Int64')
```
